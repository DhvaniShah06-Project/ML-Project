```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
```

```python
df = pd.read_csv('../../ML Lab/Position_Salaries.csv')
df
```

```python
x = df[['Level']]
y = df['Salary']
```

```python
model = LinearRegression()
model.fit(x,y)
print('Slope (m) : ',model.coef_)
print('Intercept (c) : ',model.intercept_)
print('Prediction for Level 15 : ',model.predict([[15]]))
```

<b>Gradient Descent by Hand</b></h1>

```python
x_gradient=df['Level'].values.astype(float)
y_gradient = df['Salary'].values.astype(float)
n=len(x_gradient)
m,c = 0.0,0.0
lr = 0.001
epochs = 20000

loss_history = []
for i in range(epochs):
    y_pred = m*x_gradient + c
    error = y_pred - y_gradient
    
    loss = (1/n) * np.sum(error ** 2)
    loss_history.append(loss)
    
    dm = (2/n) * np.sum(error*x_gradient)
    dc = (2/n) * np.sum(error)
    
    m = m-lr*dm
    c=c-lr*dc
    
    if i%4000 ==0:
        print(f'epoch {i}: loss={loss:.2f} m={m:.3f} c={c:.3f}')
        
print('Final Parameters -> m:',m,'c:',c)
```
