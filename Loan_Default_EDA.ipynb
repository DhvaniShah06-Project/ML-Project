```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))


education_counts = df['Education'].value_counts()


plt.bar(education_counts.index, education_counts.values, color='lightcoral', edgecolor='black')

plt.title('Total Loans by Education Level')
plt.xlabel('Education Level')
plt.ylabel('Number of Loans')
plt.grid(axis='y', linestyle=':', alpha=0.6)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))

plt.scatter(df['Income'], df['LoanAmount'], color='purple', alpha=0.4, edgecolors='none')

plt.title(' Income vs. Requested Loan Amount')
plt.xlabel('Annual Income ')
plt.ylabel('Loan Amount ')
plt.grid(True, linestyle=':', alpha=0.6)
plt.show()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))


grouped_data = df.groupby('InterestRate')['LoanAmount'].mean()

plt.bar(grouped_data.index, grouped_data.values, color='blue', alpha=0.6, width=0.4)

plt.title('Interest Rate vs. Average Loan Amount')
plt.xlabel('Interest Rate')
plt.ylabel('Average Loan Amount')
plt.grid(True, linestyle=':', alpha=0.6)

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 5))

# Seaborn handles huge datasets much faster for categorical plots
sns.boxplot(data=df, x='Default', y='Income', palette='Set2')

plt.title('Income Distribution by Default Status')
plt.xlabel('Default Status (0 = No, 1 = Yes)')
plt.ylabel('Income')
plt.grid(True, linestyle=':', alpha=0.6, axis='y')

plt.tight_layout()
plt.show()

```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))

# Aggregate data first so Matplotlib only draws a few summary bars
purpose_counts = df['LoanPurpose'].value_counts()

plt.bar(purpose_counts.index, purpose_counts.values, color='teal', alpha=0.7, edgecolor='black')

plt.title('Number of Loans by Purpose')
plt.xlabel('Loan Purpose')
plt.ylabel('Count')
plt.grid(True, linestyle=':', alpha=0.6, axis='y')

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))

counts, bins = np.histogram(df['DTIRatio'].dropna(), bins=40)

plt.stairs(counts, bins, fill=True, color='purple', alpha=0.6, edgecolor='black')

plt.title('Distribution of Debt-to-Income (DTI) Ratio')
plt.xlabel('DTI Ratio')
plt.ylabel('Count')
plt.grid(True, linestyle=':', alpha=0.6)

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))

age_grouped = df.groupby('Age')['LoanAmount'].mean()

plt.plot(age_grouped.index, age_grouped.values, color='red', linewidth=2, marker='o', markersize=4)

plt.title('Average Loan Amount by Age Group')
plt.xlabel('Age')
plt.ylabel('Average Loan Amount')
plt.grid(True, linestyle=':', alpha=0.6)

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(7, 7))

education_counts = df['Education'].value_counts()

plt.pie(education_counts.values, labels=education_counts.index, autopct='%1.1f%%', startangle=140,  colors=['#4f81bd', '#c0504d', '#9bbb59', '#8064a2'])

plt.title('Percentage of Loan Taken by people on the basis of Education Level')

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(7, 7))

yes_no_Counts = df['Default'].value_counts()

plt.pie(yes_no_Counts.values, labels=yes_no_Counts.index, autopct='%1.1f%%', startangle=140,  colors=['#4f81bd', '#c0504d'])

plt.title('Percentage of Default Loan Payment')

plt.tight_layout()
plt.show()
# 0 is no and 1 is yes
```
